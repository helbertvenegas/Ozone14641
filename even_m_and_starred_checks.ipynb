This notebook verifies the finite abelian group **conditional on** Theorem B, Table `tab:split`, and Propositions `prop:MNP` and `prop:OWZ` of the revised monomial manuscript. It does **not** prove those input claims from the Zhang–Zhang matrices.

Assume the algebra is PI and $m=\operatorname{ord}(u^2)$ is even, with $m'=\operatorname{ord}(v^2)$. Then

$$\operatorname{Oz}(B)\cong C_2\times C_{2m'}^2$$

for all six families $B=\mathbb M,\mathbb N,\mathbb O,\mathbb P,\mathbb W,\mathbb Z$.


```python
import sympy as sp
from sympy.matrices.normalforms import smith_normal_form
from sympy.polys.domains import ZZ
families = list("MNOPWZ")
def smith(m, mp):
    assert m > 0 and m % 2 == 0 and mp > 0
    relation_matrix = sp.diag(2*mp, 2*mp, 2)
    D = smith_normal_form(relation_matrix, domain=ZZ)
    return tuple(sorted(abs(int(D[j,j])) for j in range(3)))
for family in families:
    for m in (2,4,6,8,10):
        for mp in (1,2,3,4,5,8):
            assert smith(m,mp) == (2,2*mp,2*mp)
print("PASS: 6 families × 5 even m × 6 mprime = 180 exact checks")
```
